```python
import random
import string


def random_letter_sequences(count: int, length: int = 5) -> list[str]:
    """Return `count` random lowercase letter sequences."""
    return [
        "".join(random.choices(string.ascii_lowercase, k=length))
        for _ in range(count)
    ]
```

```python
def test_random_letter_sequences():
    count, length = 4, 5
    random.seed(0)
    sequences = random_letter_sequences(count, length)
    expected = ["vtkgn", "kuhmp", "xnhtq", "gxzvx"]
    print("input:", count, length)
    print("output:", sequences)
    assert sequences == expected
    print("ok")


test_random_letter_sequences()
```
